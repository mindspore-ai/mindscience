## Frontierscience Olympiad Chemistry 测试任务工作流演示

本案例演示如何使用 MindScienceAgent 进行化学奥林匹克竞赛问题的自动化求解测试。

### 1. 环境设置与配置加载

首先，添加项目路径并导入必要的模块。

```python
import sys
sys.path.append("../../")

from mindscience_agent.workflow import ExperimentWorkflow
from mindscience_agent.config import MindScienceConfig
```

### 2. 加载配置文件

加载实验配置文件，可以根据实际情况修改，**配置文件中需包含用户自己的模型 API 密钥等信息**。

```python
config_path = "./mindscience_agent.yaml"
config = MindScienceConfig.init_config_from_yaml(config_path)
```

### 3. 初始化实验工作流

创建实验工作流实例，配置相关参数。

```python
# init workflow
agent = ExperimentWorkflow(
    config=config,
    enable_critic=False,
    test_time_scale_round=1
)
```

### 4. 定义研究目标

在prompt中描述研究问题，Agent 将自动解析并规划工作流。本案例想要解决一个化学奥林匹克竞赛问题：一当量的2,3,4,5-四甲基-己二酸根进行电解，与自身反应生成主要的对称产物X。确定分子X。

```python
prompt = """One equivalent of &lt;INCHI&gt;InChI=1S/C10H18O4/c1-5(7(3)9(11)12)6(2)8(4)10(13)14/h5-8H,1-4H3,(H,11,12)(H,13,14)/p-2&lt;/INCHI&gt;, &lt;SMILES&gt;CC(C(C)C(C)C(\[O-\])=O)C(C)C(\[O-\])=O&lt;/SMILES&gt;, &lt;IUPAC&gt;2,3,4,5-tetramethyl-hexanedioate&lt;/IUPAC&gt; undergoes electrolysis, reacting with itself to give the major symmetric product X. Identify molecule X."""
```

### 5. 自动化执行工作流

运行 Agent 工作流，自动执行问题分析并使用chemistry-query skill去解决该问题，最终输出包裹在<solution>标签中的答案。

```python
# run workflow
result = await agent.run(prompt)
```
